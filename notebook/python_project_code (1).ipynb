```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv("/content/2017.csv")

# Display first 5 rows
print("First 5 Rows:")
display(df.head())

# Display last 5 rows
print("Last 5 Rows:")
display(df.tail())

# Dataset shape
print("Dataset Shape:", df.shape)

# Dataset information
print("\nDataset Information:")
df.info()

# Statistical description
print("\nDataset Description:")
display(df.describe())

# Check missing values in each column
print("Missing Values:")
print(df.isnull().sum())

# Total missing values
print("\nTotal Missing Values:")
print(df.isnull().sum().sum())

# Count duplicate rows
print("Number of Duplicate Rows:", df.duplicated().sum())

# Sort countries by happiness score
top_10 = df.sort_values(
    by="Happiness.Score",
    ascending=False
).head(10)

print("Top 10 Happiest Countries:")
display(top_10[["Country", "Happiness.Score"]])

# Sort countries by happiness score in ascending order
bottom_10 = df.sort_values(
    by="Happiness.Score",
    ascending=True
).head(10)

print("10 Least Happy Countries:")
display(bottom_10[["Country", "Happiness.Score"]])

# Extract happiness scores as NumPy array
scores = df["Happiness.Score"].to_numpy()

# Calculate statistical values using NumPy
mean_score = np.mean(scores)
median_score = np.median(scores)
min_score = np.min(scores)
max_score = np.max(scores)

print("Happiness Score Statistics")
print("--------------------------")
print("Mean Score:", mean_score)
print("Median Score:", median_score)
print("Minimum Score:", min_score)
print("Maximum Score:", max_score)

if "Regional indicator" in df.columns:
    region_avg = df.groupby(
        "Regional indicator"
    )["Happiness.Score"].mean().sort_values(ascending=False)

    print("Average Happiness Score by Region:")
    display(region_avg)

    # Region-wise bar chart
    region_avg.plot(
        kind="bar",
        figsize=(12, 6),
        color="skyblue"
    )
    plt.title("Average Happiness Score by Region")
    plt.xlabel("Region")
    plt.ylabel("Average Happiness Score")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()
else:
    print("Regional indicator column is not available in this dataset.")

# Top 10 countries by GDP per capita (Economy..GDP.per.Capita. in 2017 dataset)
top_gdp = df.sort_values(
    by="Economy..GDP.per.Capita.",
    ascending=False
).head(10)

print("Top 10 Countries by GDP per Capita:")
display(top_gdp[["Country", "Economy..GDP.per.Capita."]])

# Top 10 countries by Family (used as an analog for Social Support in 2017 dataset)
top_social = df.sort_values(
    by="Family",
    ascending=False
).head(10)

print("Top 10 Countries by Social Support (Family):")
display(top_social[["Country", "Family"]])

# Bar plot for Top 10 Happiest Countries
plt.figure(figsize=(12, 6))
sns.barplot(
    data=top_10,
    x="Country",
    y="Happiness.Score",
    color="skyblue"
)
plt.title("Top 10 Happiest Countries in 2017")
plt.xlabel("Country")
plt.ylabel("Happiness Score")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()

# Scatter plot for GDP vs Happiness
plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=df,
    x="Economy..GDP.per.Capita.",
    y="Happiness.Score",
    color="green",
    s=70
)
plt.title("GDP per Capita vs Happiness Score")
plt.xlabel("GDP per Capita")
plt.ylabel("Happiness Score")
plt.tight_layout()
plt.show()

print("OBSERVATIONS")
print("------------")

# Observation 1: Happiest country
happiest = df.loc[df["Happiness.Score"].idxmax()]
print(
    "1. The happiest country is",
    happiest["Country"],
    "with a happiness score of",
    happiest["Happiness.Score"]
)

# Observation 2: Least happy country
least_happy = df.loc[df["Happiness.Score"].idxmin()]
print(
    "2. The least happy country is",
    least_happy["Country"],
    "with a happiness score of",
    least_happy["Happiness.Score"]
)

# Observation 3: GDP and happiness relationship
correlation = df["Economy..GDP.per.Capita."].corr(df["Happiness.Score"])
print(
    "3. The correlation between GDP per capita and happiness score is",
    round(correlation, 3)
)

if correlation > 0:
    print("   GDP per capita has a positive relationship with happiness.")
elif correlation < 0:
    print("   GDP per capita has a negative relationship with happiness.")
else:
    print("   GDP per capita has no linear relationship with happiness.")
```
